```python
import pandas as pd
import numpy as np

# Data wrangling, preparation, and sampling

# Load the raw dataset
df = pd.read_csv("clean_sheet.csv")

# Only include relevant columns
df = df[['Player', 'Squad', 'MP', 'Min', 'Save%']].copy()

# Filter out backup players with <90 minutes played
df = df[df['Min'] >= 90]

# Clean the target variable by converting it to numbers and dropping blanks
df['Save%'] = pd.to_numeric(df['Save%'], errors='coerce')
df = df.dropna(subset=['Save%'])

# Feature Engineering: Label players by Tournament Stage
df['Team_Max_Matches'] = df.groupby('Squad')['MP'].transform('max')
df['Stage'] = np.where(df['Team_Max_Matches'] > 3, 'Knockout Stage', 'Group Stage')

# Split into our two groups
df_knockout = df[df['Stage'] == 'Knockout Stage']
df_group = df[df['Stage'] == 'Group Stage']

# Set a random state for reproducibility
RANDOM_STATE = 12345

# Take random samples with n = 25 (since the Group Stage only has 25 total valid keepers)
sample_ko = df_knockout.sample(n=25, random_state=RANDOM_STATE)
sample_gs = df_group.sample(n=25, random_state=RANDOM_STATE)

# Write samples to a CSV file (optional, if you need to export them)
sample_ko.to_csv("sample_knockout_keepers.csv", index=False)
sample_gs.to_csv("sample_group_keepers.csv", index=False)

# Display a quick check
print("Knockout Sample:")
display(sample_ko.head())
print("\nGroup Stage Sample:")
display(sample_gs.head())
```
